TRATAMENTO DE DADOS, COMEÇANDO PELA "APARENCIA"

```python
dataset['Aparencia'].describe()
```

```python
#Não pode ter dados diferente de "SoL", "Nublado", "Chuva"
#calcular a mediana

agrupar = dataset.groupby(['Aparencia']).size()
agrupar
```

```python
# Preencher com o valor que mais aparece
# e mudar nome da coluna

dataset.loc[dataset['Aparencia'] == 'sol','Aparencia'] = "Sol"
dataset.loc[dataset['Aparencia'] == 'chuva', 'Aparencia'] = 'Chuva'
dataset.loc[dataset['Aparencia'] == 'nublado', 'Aparencia'] = 'Nublado'
```

```python
agrupado = dataset.groupby(['Aparencia']).size()
agrupado
```

TRATAMENTO DE DADOS "TEMPERATURA"

```python
dataset["Temperatura"].describe()
```

```python
#Visualizar

dataset.loc[(dataset['Temperatura'] <= -130) | (dataset['Temperatura'] >= 130)]
```

```python
#calcular a mediana

mediana = sts.median(dataset['Temperatura'])
mediana
```

```python
#substituir os valore fora do range pela mediana

dataset.loc[(dataset['Temperatura'] < -130) | (dataset['Temperatura'] > 130), 'Temperatura'] = int(mediana)
```

```python
dataset["Temperatura"].describe()
```
